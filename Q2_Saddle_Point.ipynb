Saddle-point plots and checks.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
OUT = Path("figures")
OUT.mkdir(exist_ok=True)
```

## Q2(a)(iv) — Left saddle inequality

```python
mu = np.linspace(0,5,400)
L = np.ones_like(mu)

fig, ax = plt.subplots(figsize=(6.2,3.8))
ax.plot(mu,L,linewidth=2.2,label=r'$L(\\mu,x^*=1)=1$')
ax.scatter([2],[1],s=65,zorder=5,label=r'$L(\\mu^*,x^*)=1$')
ax.set_xlabel(r'Dual variable $\\mu$')
ax.set_ylabel(r'$L(\\mu,1)$')
ax.set_title('Left saddle inequality')
ax.set_xlim(0,5)
ax.set_ylim(0.72,1.28)
ax.grid(True,alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUT/'Q2a_LeftSaddle.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Q2(a)(iv) — Right saddle inequality

```python
x = np.linspace(-1,3,500)
L = (x-1)**2 + 1

fig, ax = plt.subplots(figsize=(6.2,3.8))
ax.plot(x,L,linewidth=2.2,label=r'$L(\\mu^*=2,x)=(x-1)^2+1$')
ax.scatter([1],[1],s=65,zorder=5,label=r'$L(\\mu^*,x^*)=1$')
ax.axhline(1,linestyle='--',linewidth=1.2)
ax.set_xlabel(r'Primal variable $x$')
ax.set_ylabel(r'$L(2,x)$')
ax.set_title('Right saddle inequality')
ax.set_xlim(-1,3)
ax.set_ylim(0.7,5.4)
ax.grid(True,alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUT/'Q2a_RightSaddle.png', dpi=300, bbox_inches='tight')
plt.show()
```
