Weak duality and lower-envelope plot.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
OUT = Path("figures")
OUT.mkdir(exist_ok=True)
```

## Q3(a) — Numerical weak-duality check

```python
def d(mu):
    return mu - mu**2/4

p_star = 1
for mu in [0,1,2,3]:
    value = d(mu)
    print(f"mu={mu}, d(mu)={value}, d(mu)<=p*: {value <= p_star}")
```

## Q3(b) — Lower envelope

```python
mu = np.linspace(0,5,600)
L0 = mu
L1 = np.ones_like(mu)
L2 = 4-mu
lower = np.minimum.reduce([L0,L1,L2])
dual = mu - mu**2/4

fig, ax = plt.subplots(figsize=(8.8,5.3))
ax.plot(mu,L0,linewidth=1.8,label=r'$x=0:\\ L(\\mu,0)=\\mu$')
ax.plot(mu,L1,linewidth=1.8,label=r'$x=1:\\ L(\\mu,1)=1$')
ax.plot(mu,L2,linewidth=1.8,label=r'$x=2:\\ L(\\mu,2)=4-\\mu$')
ax.plot(mu,lower,linewidth=3.0,label=r'Lower envelope $\\tilde d(\\mu)$')
ax.plot(mu,dual,linestyle='--',linewidth=2.0,label=r'Exact dual $d(\\mu)=\\mu-\\mu^2/4$')
ax.scatter([1,3],[1,1],s=55,zorder=5)
ax.scatter([2],[1],s=65,zorder=6)
ax.annotate(r'$\\mu^*=2$', xy=(2,1), xytext=(2.15,0.55),
            arrowprops=dict(arrowstyle='->',linewidth=0.9), fontsize=10)
ax.set_xlabel(r'Dual variable $\\mu$')
ax.set_ylabel('Function value')
ax.set_title('Lagrangian lines and lower envelope')
ax.set_xlim(0,5)
ax.set_ylim(-1.35,5.2)
ax.grid(True,alpha=0.25)
ax.legend(loc='upper right',fontsize=8.5)
fig.tight_layout()
fig.savefig(OUT/'Q3b_LowerEnvelope.png', dpi=300, bbox_inches='tight')
plt.show()

print("Lower envelope is concave.")
print("min(lower - dual) =", np.min(lower-dual))
```
