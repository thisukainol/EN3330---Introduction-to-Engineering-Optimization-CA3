```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
OUT = Path("figures")
OUT.mkdir(exist_ok=True)
```

## Q1(a) — Lagrangian curves and minimizers

```python
x = np.linspace(-1.2, 3.2, 800)
mus = [0, 1, 2, 3]

fig, ax = plt.subplots(figsize=(9, 5.8))
for mu in mus:
    L = x**2 + mu*(1-x)
    ax.plot(x, L, linewidth=2, label=fr'$\\mu={mu}$')
    xm = mu/2
    Lm = xm**2 + mu*(1-xm)
    ax.scatter([xm], [Lm], s=55, zorder=5)
    ax.annotate(fr'$({xm:g},\\,{Lm:g})$', (xm,Lm),
                xytext=(7, 10 if mu != 1 else -20),
                textcoords='offset points', fontsize=9)

ax.axvline(1, linestyle='--', linewidth=1.4, label=r'Constraint boundary $x=1$')
ax.scatter([1],[1], s=70, zorder=6)
ax.set_xlabel(r'Primal variable $x$')
ax.set_ylabel(r'$L(\\mu,x)$')
ax.set_title('Lagrangian curves and their minimizers')
ax.set_xlim(-1.2, 3.2)
ax.grid(True, alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUT/'Q1a_Lagrangian.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Q1(b) — Dual function and dual optimum

```python
mu = np.linspace(0,5,600)
d = mu - mu**2/4

fig, ax = plt.subplots(figsize=(8.6,5.2))
ax.plot(mu,d,linewidth=2.4,label=r'$d(\\mu)=\\mu-\\mu^2/4$')
ax.scatter([2],[1],s=70,zorder=5)
ax.axvline(2,linestyle='--',linewidth=1.2)
ax.axhline(1,linestyle='--',linewidth=1.2,label=r'$d^*=1$')
ax.annotate(r'$(\\mu^*,d^*)=(2,1)$', xy=(2,1), xytext=(2.55,0.62),
            arrowprops=dict(arrowstyle='->',linewidth=1.0), fontsize=11)
ax.set_xlabel(r'Dual variable $\\mu$')
ax.set_ylabel(r'Dual function $d(\\mu)$')
ax.set_title('Dual function and dual optimum')
ax.set_xlim(0,5)
ax.set_ylim(-1.35,1.12)
ax.grid(True,alpha=0.25)
ax.legend(loc='lower left')
fig.tight_layout()
fig.savefig(OUT/'Q1b_DualFunction.png', dpi=300, bbox_inches='tight')
plt.show()

print("mu*=2, d*=1")
```

## Q1(c) — Primal function

```python
x = np.linspace(-1,3,800)
Ms = [2,6,15]

fig, ax = plt.subplots(figsize=(8.8,5.4))
for M in Ms:
    gM = x**2 + M*np.maximum(1-x,0)
    ax.plot(x,gM,linestyle='--',linewidth=1.8,
            label=fr'$g_{{{M}}}(x)=\\max_{{0\\leq\\mu\\leq {M}}}L(\\mu,x)$')

xf = np.linspace(1,3,400)
ax.plot(xf,xf**2,linewidth=2.7,label=r'Exact finite branch: $g(x)=x^2,\\ x\\geq1$')
ax.axvline(1,linestyle=':',linewidth=1.5,label=r'Constraint boundary $x=1$')
ax.scatter([1],[1],s=65,zorder=5)
ax.annotate(r'$(x^*,g(x^*))=(1,1)$', xy=(1,1), xytext=(1.32,2.05),
            arrowprops=dict(arrowstyle='->',linewidth=1.0), fontsize=10)
ax.set_xlim(-1,3)
ax.set_ylim(0,11)
ax.set_xlabel(r'Primal variable $x$')
ax.set_ylabel(r'$g_M(x)$ or $g(x)$')
ax.set_title('Primal function and finite-multiplier approximations')
ax.grid(True,alpha=0.25)
ax.legend(loc='upper right',fontsize=8.5)
fig.tight_layout()
fig.savefig(OUT/'Q1c_PrimalFunction.png', dpi=300, bbox_inches='tight')
plt.show()
```
